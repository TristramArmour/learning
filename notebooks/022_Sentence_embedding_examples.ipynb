# Creating my own test lab for sentence transformers

```python
from sentence_transformers import SentenceTransformer, util
import torch

```

```python
corpus = ["The lion lives in the zoo",
          "Most lions live on the savannah",
          "King of the jungle",
         "I like turtles"]
```

Small but fast model - 'all-MiniLM-L6-v2' has 384 dimensions

```python
# download the sentence embeddings model
embedder = SentenceTransformer('all-MiniLM-L6-v2')
```

```python
# embed the text
corpus_embeddings = embedder.encode(corpus, convert_to_tensor=True)
print(corpus_embeddings.shape)
```

```python
query = 'Where does the lion live?'
# Turn the content into embedded vector
query_embedding = embedder.encode(query, convert_to_tensor=True)
# retrieve cosine scores
cos_scores = util.cos_sim(query_embedding, corpus_embeddings)[0]
```

```python
for score, text in zip(cos_scores, corpus):
    print(f"Similarity score: {score} for text: {text}")
```

Larger model - gte-large-en-v1.5 - 1024 dimensions

```python
# download the sentence embeddings model
embedder = SentenceTransformer('Alibaba-NLP/gte-large-en-v1.5',trust_remote_code=True)
```

```python
# embed the text
corpus_embeddings = embedder.encode(corpus, convert_to_tensor=True)
print(corpus_embeddings.shape)
```

```python
query = 'Where does the lion live?'


# Turn the content into embedded vector
query_embedding = embedder.encode(query, convert_to_tensor=True)
# retrieve cosine scores
cos_scores = util.cos_sim(query_embedding, corpus_embeddings)[0]
for score, text in zip(cos_scores, corpus):
    print(f"Similarity score: {score} for text: {text}")
```
